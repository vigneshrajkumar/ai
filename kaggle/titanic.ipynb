```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn import tree
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

## Data Cleanup
- Check nature of the columns - Nominal, Ordinal, Ratio, Integer
- Check for missing values

## Try 1 - Logistic Regression - Single Feature

- Prediction based on only age of the passenger
- Data points where age is missing, the median value of age is used. - **62% accuracy** in Kaggle
- `X` are called features and `Y` is called target

```python
training_set = pd.read_csv("./../data/titanic/train.csv")

X_train = training_set[["Age"]] 

# shows the number of nulls per feature
print(training_set.isnull().sum())

# handling missing age data in train data
median_age = X_train['Age'].median()
print(f"median age: {median_age}")
X_train['Age'] = X_train['Age'].fillna(median_age)

# model training with train data
clf = LogisticRegression()
Y_train = training_set[["Survived"]]
clf.fit(X_train, Y_train)

accuracy = clf.score(X_train, Y_train)
print(f"Model Accuracy: {accuracy:.4f}")


# publishing results with test data

test_set = pd.read_csv("./../data/titanic/test.csv")

X_test = test_set[["Age",  "PassengerId"]]

# handling missing age data in test data
median_age = X_test['Age'].median()
print(f"median age: {median_age}")
X_test['Age'] = X_test['Age'].fillna(median_age)


model_input = X_test[["Age"]]
predictions = clf.predict(model_input)

results = pd.DataFrame({
    "PassengerId": X_test["PassengerId"],
    "Survived": predictions
})

results.to_csv("./../data/titanic/titanic_results.csv", header=True, index = False)
```

## Try 2 - Logistic Regression with multiple features

- Modelled with features Sex and Age - **Accuracy is 77%** in Kaggle
- Age has been scaled.  
- `clf.fit()` will automatically assign weights to each feature while minimizing the overall cost function

| Method                         | Accuracy in Kaggle |
| -                              | -                  |
| Sex and Age as features        | 77%                |
| Scaled Sex and Age as features | 72%                |

```python
training_set = pd.read_csv("./../data/titanic/train.csv")

X_train = training_set[["Age", "Sex"]]

# handling missing data in train data
sex_map = {"male": 0, "female": 1}
X_train["Sex"] = X_train["Sex"].map(sex_map)

median_age = X_train['Age'].median()
print(f"median age: {median_age}")
X_train['Age'] = X_train['Age'].fillna(median_age)

# model training with train data
clf = LogisticRegression()
Y_train = training_set[["Survived"]]

pipe = Pipeline([
    # ("scaler", StandardScaler()), # scaling dropped accuracy by 5%
    ("clf", LogisticRegression())
])

pipe.fit(X_train, Y_train)

accuracy = pipe.score(X_train, Y_train)
print(f"Model Accuracy: {accuracy:.4f}")


# publishing results with test data
test_set = pd.read_csv("./../data/titanic/test.csv")
X_test = test_set[["Age",  "PassengerId", "Sex"]]

# handling missing data in test data
median_age = X_test['Age'].median()
print(f"median age: {median_age}")
X_test['Age'] = X_test['Age'].fillna(median_age)
X_test["Sex"] = X_test["Sex"].map(sex_map)

model_input = X_test[["Age", "Sex"]]
predictions = pipe.predict(model_input)

results = pd.DataFrame({
    "PassengerId": X_test["PassengerId"],
    "Survived": predictions
})

results.to_csv("./../data/titanic/titanic_results.csv", header=True, index = False)
```

## Try 3 - Decision Tree

- Modelled with features Sex and Age - **Accuracy is 72%** in Kaggle

```python
training_set = pd.read_csv("./../data/titanic/train.csv")

X_train = training_set[["Age", "Sex"]]

# handling missing data in train data
sex_map = {"male": 0, "female": 1}
X_train["Sex"] = X_train["Sex"].map(sex_map)

median_age = X_train['Age'].median()
print(f"median age: {median_age}")
X_train['Age'] = X_train['Age'].fillna(median_age)

# model training with train data
clf = tree.DecisionTreeClassifier()
Y_train = training_set[["Survived"]]
clf.fit(X_train, Y_train)

accuracy = clf.score(X_train, Y_train)
print(f"Model Accuracy: {accuracy:.4f}")


# publishing results with test data

test_set = pd.read_csv("./../data/titanic/test.csv")

X_test = test_set[["Age",  "PassengerId", "Sex"]]

# handling missing data in test data
median_age = X_test['Age'].median()
print(f"median age: {median_age}")
X_test['Age'] = X_test['Age'].fillna(median_age)
X_test["Sex"] = X_test["Sex"].map(sex_map)

model_input = X_test[["Age", "Sex"]]
predictions = clf.predict(model_input)

results = pd.DataFrame({
    "PassengerId": X_test["PassengerId"],
    "Survived": predictions
})

results.to_csv("./../data/titanic/titanic_results.csv", header=True, index = False)
```
