### Try 1: Single Feature

- `LotArea` as the only feature, **RMSE of 59,453** in Kaggle
- No Concept of accuracy in the model like LogisticRegression (it had the final output unlike Linear Regression)

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
full_training_data = pd.read_csv("./../data/house_price/train.csv")

X_train = full_training_data[["LotArea"]]
# print(X_train.head())

Y_train = full_training_data["SalePrice"]
# print(Y_train.head())

reg = LinearRegression()
reg.fit(X_train, Y_train)

full_test_data = pd.read_csv("./../data/house_price/test.csv")

X_test = full_test_data[["LotArea"]]

results = pd.DataFrame({
    "SalePrice": reg.predict(X_test),
    "Id": full_test_data["Id"]
})

results.to_csv("./../data/house_price/results.csv", index=False)
```
